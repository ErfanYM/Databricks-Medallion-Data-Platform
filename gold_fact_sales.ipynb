# Trasnformation Logic

```python
query = """
SELECT 
  sd.order_number,
  pr.product_key,
  cu.customer_key,
  sd.order_date,
  sd.ship_date,
  sd.due_date,
  sd.sales_amount,
  sd.quantity,
  sd.price
FROM silver.crm_sales sd
left join workspace.gold.dim_products pr
    on sd.product_number = pr.product_number
left join workspace.gold.dim_customers cu
    on sd.customer_id = cu.customer_id
"""

df = spark.sql(query) 
```

```python
df.limit(10).display()
```

# Writing Gold table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.fact_sales")
```
